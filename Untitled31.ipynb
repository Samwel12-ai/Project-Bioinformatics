```r
# Step 2: Create a DNA string object


dna_sample <- DNAString("ATGCGTACGTTAGC")
print(dna_sample)
```

```r
# Step 1: Install and load necessary packages


if (!requireNamespace("BiocManager", quietly = TRUE)) {
  install.packages("BiocManager")
}

BiocManager::install("Biostrings")
install.packages("tidyverse")

library(Biostrings)
library(tidyverse)
```

```r
# Step 3: Translate the DNA sequence into protein


protein_output <- translate(dna_sample)
print(protein_output)
```

```r
# Step 4: Get the reverse complement of the DNA sequence


dna_reverse <- reverseComplement(dna_sample)
dna_reverse
```

```r
# Step 5: Calculate GC content percentage


dna_seq <- DNAString("ATGCGTACGTTAGC")

# GC count
gc_count <- letterFrequency(dna_seq, c("G","C"))

# Length
len <- nchar(as.character(dna_seq))

# Percentage
gc_percent <- sum(gc_count) / len * 100

gc_percent
```

```r
# Step 6: Create the gene data frame


genes_tbl <- tibble(
  gene_id = c("G1", "G2", "G3"),
  expr_value = c(10, 50, 30),
  category = c("A", "B", "A")
)
```

```r
# Step 7: Filter out significant genes


filtered_tbl <- genes_tbl %>%
  dplyr::filter(expr_value >= 25)
```

```r
# Step 8: Create a new column for log expression


genes_tbl <- genes_tbl %>%
  mutate(log_val = log(expr_value))
```

```r
# Step 9: Sort genes by log expression


sorted_tbl <- genes_tbl %>%
  arrange(desc(log_val))
```

```r
# Step 10: Create a DNA sequence set

seq_collection <- DNAStringSet(c("ATGCGT", "TTGCAA", "GGGCCC"))
```

```r
# Step 11: Convert sequence set to a data frame


seq_table <- tibble(seq_text = as.character(seq_collection))
```

```r
# Step 11: Convert sequence set to a data frame


final_dataset <- bind_cols(genes_tbl, seq_table)
```

```r
# Step 13: Summarize mean expression by group


group_summary <- genes_tbl %>%
  group_by(category) %>%
  summarise(avg_expr = mean(expr_value))
```
